```python
#1
def reverse_string(s):
    if len(s) == 0:
        return s
    else:
        return reverse_string(s[1:]) + s[0]
result = reverse_string("hello")
print("Reversed String:", result)
```

```python
#2
def sum_playlist_durations(durations):
    if len(durations) == 0:
        return 0
    return durations[0] + sum_playlist_durations(durations[1:])
durations = [180, 240, 150, 210]
total = sum_playlist_durations(durations)
print("Total playlist duration:", total, "seconds")
```

```python
#3
count = 10
def update_count():
    count = 5
    print('Inside:', count)
update_count()
print('Outside:', count)
```

```python
#4
def count_likes(posts):
    total = posts.get('likes', 0)
    for reply in posts.get('replies', []):
        total += count_likes(reply)
    return total
posts = {'likes': 100,'replies': [{'likes': 50,'replies': [{'likes': 20, 'replies': []},{'likes': 10, 'replies': []}]},{'likes': 30,'replies': [{'likes': 15, 'replies': []}]}]}
print("Total likes:", count_likes(posts))
```
