```python
#2
score=[45,78,10,34,127,89,102]
i=0
while i < len(score):
  if score[i]>100:
    break
  else:
    print(score[i])
  i+=1
```

```python
#3
item_prices = [299, 499, 199, 999, 149]
total = 0
for price in item_prices:
    if price < 200:
        continue
    total = total + price
    print("Included Price:", price)
print("Total Cart Value:", total)
```

```python
#5
followers = [120, 1500, 23000, 800, 45000]
for count in followers:
    if count < 1000:
        print(count, "-", "Micro")
    elif count <= 10000:
        print(count, "-", "Influencer")
    else:
        print(count, "-", "Celebrity")
```
