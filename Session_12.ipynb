```python
#1
gst_calculator = lambda price: price + (price * 0.18)
print("Price after GST on Rs. 100:", gst_calculator(100))
print("Price after GST on Rs. 250:", gst_calculator(250))
print("Price after GST on Rs. 500:", gst_calculator(500))
```

```python
#2
songs = [' shape OF you ', '  kesariya', 'BELIEVER ',
         ' blinding LIGHTS ', ' EXCUSES  ']
cleaned_songs = list(map(lambda song: song.strip().title(), songs))
print(cleaned_songs)
```

```python
#3
products = ['Samsung TV', 'iPhone 14', 'Shoes',
            'Smart Watch', 'Laptop', 'Speaker', 'sunglasses']
filtered_products = list(filter(lambda product: product.lower().startswith('s'), products))
print(filtered_products)
```

```python
#4
from functools import reduce
order_amounts = [120, 340, 560, 80]
total_bill = reduce(lambda a, b: a + b, order_amounts)
print("Total Zomato Bill: Rs.", total_bill)
```

```python
#5
from functools import reduce
numbers = [40, 60, 80, 120]
doubled = list(map(lambda x: x * 2, numbers))
filtered = list(filter(lambda x: x > 100, doubled))
total = reduce(lambda x, y: x + y, filtered)
print("Original List:", numbers)
print("Doubled Numbers:", doubled)
print("Filtered Numbers:", filtered)
print("Final Sum:", total)
```
