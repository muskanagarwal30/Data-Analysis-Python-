```python
#2
def clean_brand_name(name):
  name=name.strip()
  name=name.replace("-"," ")
  return name
```

```python
brand_name=clean_brand_name("  oneplus-Nord  ")
print(brand_name)
```

```python
#4
def format_product_display(name, price):
  return f"{name} - Rs.{price}"

print("Formated Product Name:",format_product_display("Boat Earbuds",1299))
```
